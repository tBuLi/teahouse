```python
from kingdon import Algebra

alg = Algebra(3, 0, 1)
globals().update(alg.blades)
```

The first line is Kingdon initialization code; the `alg = Algebra(3, 0, 1)` line creates the algebra of the desired signature; follow it with the `globals().update(alg.blades)` line to initialize and make available the basis vectors and basis blades of the algebra (e.g., `e0`, `e1`, `e12`).

The above two cells should be the first two cells of your Kingdon notebook, with the obvious variations for version numbers and signatures.

We want to make a notebook to illustrate projecting a point $A$ onto a plane $p$:  $$P_p(A) = (A \cdot p)/p.$$  We first wish to implement and test this formula in the notebook.  The Kingdon inner product operator symbol is `|`, so our formula appears in the notebook as the third line of

```python
p = (e0 + 3*e1 + 2*e2 + 3*e3).dual()
A = e3-1*e0
PpA = (A | p)/p
PpA.dual()
```

The Kingdon documention provides the <A HREF="https://kingdon.readthedocs.io/en/stable/usage.html#operators">symbols used in Kingdon for the operators</A>.  But note that in the above cell, we basically type our formulas as expected.  And the above cell isn't needed for our visualization; we only put it here to test the formula.

To visualize our formula, we are going to write some functions to create points and planes, and to create the elements we want to graph.

One way to create a function is with the `lambda` operator.  The following are the three lambda functions that we'll use in this notebook:

```python
# Construct points and planes
point = lambda x, y, z: (e0 + x*e1 + y*e2 + z*e3).dual()
plane = lambda a, b, c, d: a*e1 + b*e2 + c*e3 + d*e0
plane_thru_point = lambda n, pt: n + (pt & n)*e0
```

The format of the `lambda` function is `function-name = lambda [parameters]: function-body`.  In the above cell, we have written a function to create a PGA point from x,y,z coordinates, a function to create a PGA plane from coordinates, a function to construct a plane through a point with a particular normal.

We now write the function for our point-on-plane projection operator:

```python
# projection function
project = lambda A, p: (A | p)/p   
```

The next thing we will do is write the function that will generate the data that we want to graph.  There are two sections of the function: in the first section, we compute the values to display, and in the second section (the `return` statement below), we explicitly state what we want to display and how we want to display it:

The format of each line of the the return statement is: first, a hexadecimal value giving the color in which to display the object; second, the object to display; and third, an optional label to put on the object.

We now create the graph itself.  In the `alg.graph` statement below, the first line gives the function we want to graph; in you wanted a static graph (where the elements don't change), you could list the objects to draw here (rather than in the `return` statement of  `graph_func`).  The remaining elements specify details of how we want the graph to appear; most of these are likely self-explanatory, and you'll probably just want to cut-and-paste this cell into your own notebook and modify these parameters as desired.

You might also see a `GraphWidget` statement; this statement is generated by alg.graph, so you don't need to worry about it.

When you run the following cell, a graph will appear, illustrating the projection.  You can rotate the graph with the mouse, and you can move the points `A` and `B` with the mouse.  As you move `A`, you will see its projection `Ap` on the other plane move; as you move `B`, you will see the corresponding plane move.

```python
alg.graph(
    graph_func,
    lineWidth=3, 
    grid=1, 
    labels=1, 
    h=-0.1, 
    p=-0.15, 
    pointRadius=2,
    fontSize=2,
    scale=1,
)
```
